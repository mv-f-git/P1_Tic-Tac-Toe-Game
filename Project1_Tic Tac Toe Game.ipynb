```python
from IPython.display import clear_output
import random
```

```python
# displaying Tic Tac Toe board
def display_board(board):
    
    clear_output() #clear previous output
    
    print('   |   |')
    print(' ' + board[7] + ' | ' + board[8] + ' | ' + board[9])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[4] + ' | ' + board[5] + ' | ' + board[6])
    print('   |   |')
    print('-----------')
    print('   |   |')
    print(' ' + board[1] + ' | ' + board[2] + ' | ' + board[3])
    print('   |   |')
```

```python
# test displaying board
my_board=[' ']*10
display_board(my_board)
print(my_board)
```

```python
# testing board
test_board = ['#','X','O','X','O','X','O','X','O','X']
display_board(test_board)
```

```python
# Asking player1 to choose marker 

def player_input():
    marker = ''
    
    while not (marker == 'X' or marker == 'O'):
        marker = input('Player 1: Do you want to be X or O? ').upper()

    if marker == 'X':
        return ('X', 'O')
    else:
        return ('O', 'X')
```

```python
# function for place player's marker at selected position 
def place_marker(board, marker, position):
    board[position]=marker
    return board
```

```python
# Test place_marker function
place_marker(test_board,'$',8)
display_board(test_board)
```

```python
# check all possible winning
def win_check(board, mark):
    return ((board[1]==board[2]==board[3]==mark) or 
    (board[4]==board[5]==board[6]==mark)or
    (board[7]==board[8]==board[9]==mark)or
    (board[1]==board[4]==board[7]==mark)or
    (board[2]==board[5]==board[8]==mark)or
    (board[3]==board[6]==board[9]==mark)or
    (board[7]==board[5]==board[3]==mark)or
    (board[1]==board[5]==board[9]==mark))
```

```python
# Test win_check function 
win_check(test_board,'X')
```

```python
# choose random player to goes first
def choose_first():
    if random.randint(0,1)==0:
        return "player1"
    else:
        return "player2"
```

```python
# check selected position is available or not
def space_check(board, position):
    return board[position]==' '
```

```python
# check board is full or not
def full_board_check(board):
    for i in range(1,10):
        if space_check(board,i):
            return False
    return True
```

```python
# Ask player to chosse a position on board
def player_choice(board):
    position=0
    while not space_check(board,position)or position not in [1,2,3,4,5,6,7,8,9]:
        position=int(input("Choose a position: (1-9):"))
    return position
```

```python
# Ask player for starting new game
def replay():
    
    return input('Do you want to play again? Enter Yes or No: ').lower().startswith('y')
```

```python
# Main loop of the game:
print('Welcome to Tic Tac Toe!')

while True:
    # a new board for each game
    my_board=[' ']*10 
    
    choice=0
    current_player=' '

    # Get markers for players
    player1,player2=player_input()
    print(f'Player1 is {player1} and Player2 is {player2}')

    # Ask players if they are ready to start
    game_on=input('Are you ready to play?Yes or No?').upper()

    # display empty board
    display_board(my_board)

    
    if game_on=='YES':
        game_on=True
    else:
        game_on=False

    # randomly select the player that goes first
    turn=choose_first()
    if turn=='Player1':
        current_player='Player1'
        print('Player1 will go first')
    else:
        current_player='Player2'
        print('Player2 will go first')

    # Main game play loop
    while game_on:

        # get valid position from current player
        choice=player_choice(my_board)

        # Assign the marker based on the current player
        if current_player=='Player1':
            marker=player1
        else:
            marker=player2

        # placing marker and display updated board
        place_marker(my_board,marker, choice)
        display_board(my_board)

        # check if current player has won
        if win_check(my_board,marker)==True:
            game_on=False
            print('Congratulations! You have won the game!') 
        else:
            # switching turn if the game is not over
            if current_player=='Player1':
                current_player='Player2'
            else:
                current_player='Player1'

        # check for full board
        if full_board_check(my_board)==True:
            game_on=False
            print('Players are equal and the board is full!')
            
    # ask for playing again 
    if not replay():
        break
```
